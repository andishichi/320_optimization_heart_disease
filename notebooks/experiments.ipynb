```python
import numpy as np
from models.logistic import sigmoid, logistic_loss, logistic_grad
from models.logistic_ridge import logistic_ridge_loss, logistic_ridge_grad
```

```python
# testing logistic.py file

# testing data
X = np.array([
    [0.0, 0.0],
    [1.0, 2.0],
    [-1.0, -20.0]
])

y = np.array([0, 1, 0])

w = np.array([0.1, -0.2])

print("Scores (X @ w):", X @ w)
print("Sigmoid:", sigmoid(X @ w))   # values should be between 0 and 1
print("Loss:", logistic_loss(w, X, y))
print("Gradient:", logistic_grad(w, X, y))
```

```python
# testing logistic_ridge.py file

X1 = np.array([
    [0.0, 0.0],
    [1.0, 2.0],
    [2.0, 1.0],
    [-1.0, -2.0]
])

y = np.array([0, 1, 1, 0])

# add bias column as Last column (all 1's)
ones = np.ones((X1.shape[0], 1))
X = np.hstack([X1, ones])   # shape: (4, 3)

print("X:\n", X)
print("y:", y)
print("X shape:", X.shape)

w = np.array([0.5, -0.25, 0.1])     # [w1, w2, bias]

print("w:", w)
print("w shape:", w.shape)

base_loss = logistic_loss(w, X, y)

ridge_loss_small_alpha = logistic_ridge_loss(w, X, y, alpha=0.1)
ridge_loss_large_alpha = logistic_ridge_loss(w, X, y, alpha=10.0)

print("Plain logistic loss:", base_loss)
print("Ridge logistic loss (alpha=0.1):", ridge_loss_small_alpha)
print("Ridge logistic loss (alpha=10.0):", ridge_loss_large_alpha)
```
