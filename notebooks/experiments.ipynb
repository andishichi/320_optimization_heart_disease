```python
import numpy as np

from models.logistic import sigmoid, logistic_loss, logistic_grad, logistic_hess
from models.logistic_ridge import logistic_ridge_loss, logistic_ridge_grad, logistic_ridge_hess
from algorithms.gradient_descent import gradient_descent
from algorithms.adam import adam
from algorithms.newton import newton
from algorithms.bfgs import bfgs
from algorithms.conjugate_gradient import conjugate_gradient
```

```python
# testing array for models and algorithms

X = np.array([
    [0.2,  0.1, 1],
    [0.4,  0.3, 1],
    [0.6,  0.8, 1],
    [0.1,  1.2, 1],
    [-0.3, -0.5, 1],
    [-0.6, -0.4, 1],
    [-0.2, -1.0, 1],
    [0.3, -0.8, 1]
])

y = np.array([0, 0, 1, 1, 0, 0, 1, 1])

# initialize weights (including bias)
w0 = np.zeros(X.shape[1])
```

```python
# testing logistic.py file

w_log = np.array([0.1, -0.2, 1])

print("Scores (X @ w):", X @ w_log)
print("Sigmoid:", sigmoid(X @ w_log))   # values should be between 0 and 1
print("Loss:", logistic_loss(w_log, X, y))
print("Gradient:", logistic_grad(w_log, X, y))
```

```python
# testing logistic_ridge.py file

print("X:\n", X)
print("y:", y)
print("X shape:", X.shape)

w_ridge = np.array([0.5, -0.25, 1.0])     # [w1, w2, bias]

print("w:", w_ridge)
print("w shape:", w_ridge.shape)

base_loss = logistic_loss(w_ridge, X, y)

ridge_loss_small_alpha = logistic_ridge_loss(w_ridge, X, y, alpha=0.1)
ridge_loss_large_alpha = logistic_ridge_loss(w_ridge, X, y, alpha=10.0)

print("Plain logistic loss:", base_loss)
print("Ridge logistic loss (alpha=0.1):", ridge_loss_small_alpha)
print("Ridge logistic loss (alpha=10.0):", ridge_loss_large_alpha)
```

```python
# testing gradient_descent.py on logistic.py

print("Initial weights w0:", w0)
print("Initial loss:", logistic_loss(w0, X, y))

w_descent, history_logistic = gradient_descent(w0=w0, loss_fn=logistic_loss, grad_fn=logistic_grad, X=X, y=y, lr=0.01, iters=1000)

print("Final weights:", w_descent)
print("Number of iterations:", len(history_logistic))
print("Final logistic loss: ", history_logistic[-1])

# peeking at convergence
print("First 5 losses:", history_logistic[:5])
print("Last 5 losses:", history_logistic[-5:])
```

```python
# testing adam.py on logistic.py

print("Initial weights w0:", w0)
print("Initial loss:", logistic_loss(w0, X, y))

w_adam, history_adam = adam(w0=w0, loss_fn=logistic_loss, grad_fn=logistic_grad, X=X, y=y, lr=0.05, iters=1000)

print("Final weights:", w_adam)
print("Number of iterations:", len(history_adam))
print("Final logistic loss: ", history_adam[-1])

print("First 5 losses:", history_adam[:5])
print("Last 5 losses:", history_adam[-5:])
```

```python
# testing newton.py on logistic.py

print("Initial weights w0:", w0)
print("Initial loss:", logistic_loss(w0, X, y))

# Newton usually converges fast
w_newton, history_newton = newton(w0=w0, loss_fn=logistic_loss, grad_fn=logistic_grad, hess_fn=logistic_hess, X=X, y=y, iters=20, tol=1e-6)

print("Final weights (Newton):", w_newton)
print("Number of iterations (Newton):", len(history_newton))
print("Final logistic loss (Newton):", history_newton[-1])

print("First 5 losses (Newton):", history_newton[:5])
print("Last 5 losses (Newton):", history_newton[-5:])
```

```python
# testing bfgs.py on logistic.py

print("Initial weights w0:", w0)
print("Initial loss:", logistic_loss(w0, X, y))

w_bfgs, history_bfgs = bfgs(w0=w0, loss_fn=logistic_loss, grad_fn=logistic_grad, X=X, y=y, iters=50, tol=1e-6)

print("Final weights (BFGS):", w_bfgs)
print("Number of iterations (BFGS):", len(history_bfgs))
print("Final logistic loss (BFGS):", history_bfgs[-1])

print("First 5 losses (BFGS):", history_bfgs[:5])
print("Last 5 losses (BFGS):", history_bfgs[-5:])
```

```python
# testing conjugate_gradient.py on logistic.py

print("Initial weights w0:", w0)
print("Initial loss:", logistic_loss(w0, X, y))

w_cg, history_cg = conjugate_gradient(w0=w0, loss_fn=logistic_loss, grad_fn=logistic_grad, X=X, y=y, iters=100, tol=1e-6)

print("Final weights (CG):", w_cg)
print("Number of iterations (CG):", len(history_cg))
print("Final logistic loss (CG):", history_cg[-1])

print("First 5 losses (CG):", history_cg[:5])
print("Last 5 losses (CG):", history_cg[-5:])
```
